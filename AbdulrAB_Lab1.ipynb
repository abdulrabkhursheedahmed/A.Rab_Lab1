```python
print("========== SMART TRAVEL RECOMMENDATION AGENT ==========")

city = input("Choose your destination (Murree/Kashmir/Gilgit/Swat): ").strip().lower()
budget = int(input("Enter your travel budget (PKR): "))
travel_style = input("Enter travel style (family/adventure): ").strip().lower()

def recommend(city, budget, travel_style):
    rules = {
        "murree": {"min_budget": 20000, "style": "family", "package": "Murree Hill Station Family Package"},
        "kashmir": {"min_budget": 45000, "style": "family", "package": "Kashmir Valley Family Retreat"},
        "gilgit": {"min_budget": 55000, "style": "adventure", "package": "Gilgit Mountain Adventure Expedition"},
        "swat": {"min_budget": 35000, "style": "adventure", "package": "Swat Valley Adventure Tour"},
    }

    if city not in rules:
        return "Sorry, this destination is currently not supported."

    rule = rules[city]

    if budget >= rule["min_budget"] and travel_style == rule["style"]:
        return f" Recommended: {rule['package']}"
    elif budget < rule["min_budget"]:
        return f" Budget too low for {city.title()}. Minimum required: {rule['min_budget']} PKR."
    else:
        return f" Travel style '{travel_style}' does not match {city.title()}'s recommended style '{rule['style']}'."

result = recommend(city, budget, travel_style)
print(result)
```
